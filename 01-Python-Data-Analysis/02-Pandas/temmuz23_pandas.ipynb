```python
#Seriler
import pandas as pd
import numpy as np
a={2015:120, 2016:130, 2017:150, 2018:180}
b=(1,2,3,4,5,6,7)

series1=pd.Series(a)
series2=pd.Series(b)
series3=pd.Series([1, "Merhaba", True, 1.23])
series4=pd.Series([78,183,40,"Ozan"], index=["Kilo", "Boy", "Yaş", "İsim"]) #indexleri isimlendirdik

print(series1)
print(series2)
print(series3)
print(series4)
```

```python
liste=["Fiat", "Vw", "Opel", "Mazda","Toyata"]
indis=[i for i in range(1,6)]

series5=pd.Series(liste, index=indis)
print(series5)

print(f"5.index= {series5[5]}")
series5[5]="Hyundai"
print(f"yeni 5.index= {series5[5]}")
```

```python
#serilerde birleştirme işlemi

liste1=["Ozan", "Tunay", "Ishak", "Omer", "Ahmet"]
indis1=[i for i in range(1,6)]

liste2=["Feyza", "Ilgın", "Beyza"]
indis2=[i for i in range(1,4)]

erkekler=pd.Series(liste1, index=indis1)
kizlar=pd.Series(liste2, index=indis2)

print(pd.concat([erkekler,kizlar],ignore_index=True)) #ignore_index=çakışmaları yok saydı.

```
