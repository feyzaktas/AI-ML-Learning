```python
import matplotlib.pyplot as plt
import numpy as np
from random import randint as rnd

plt.xlabel("Yıllar") #x ekseninin etiketi
plt.ylabel("İhraacat") #y ekseninin etiketi
plt.title("Yıllara Göre İhracat rakamaları") #grafiğin genel başlığı


xpoint=np.array([i for i in range(2010, 2016)]) # x eksenindeki noktalar

ypoint=np.array([rnd(800,1200) for i in range(6)])

plt.plot(xpoint, ypoint, "P")
plt.show()

```

```python
#ÇİZGİLİ GRAFİK
plt.xlabel("Yıllar") #x ekseninin etiketi
plt.ylabel("İhraacat") #y ekseninin etiketi
plt.title("Yıllara Göre İhracat rakamaları") #grafiğin genel başlığı


xpoint=np.array([i for i in range(2010, 2016)]) # x eksenindeki noktalar

ypoint=np.array([rnd(800,1200) for i in range(6)])

plt.plot(xpoint, ypoint, marker="P") #******************önemli ksıım
##çizginin stilini (noktalı, kesik vb.) " " içinde yazarak gösteriririz.
#color: cizginin rengini belirliyoruz
#linewidth=çizginin kalınlıgı, integer girilir (lw)
#marker= grafikteki belirtilecek yerdeki simgeyi ayarlar.
#ms=simgenin büyüklügü, integer girilir (marker size)
#mec=simgenin dış rrengi
#mfc=simgenin iç rengi
# 
plt.show()
```

```python
plt.title("Accuracy Graph")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.grid()#ızgara görünümü

yTrainPoints=np.array([0.8,0.75,0.6,0.55,0.45,0.3])
yValPoints=np.array([0.8,0.65,0.6,0.5,0.6,0.7])

plt.plot(yTrainPoints, linestyle=":" , color= "m", lw="2", marker="x")
plt.plot(yValPoints, linestyle="-." , color= "b", lw="4", marker="o")

plt.legend(["Train Loss", "Val Loss"], loc=0)

plt.show()
```

```python
font1={"family":"arial", "color":"red", "size":18}
font2={"family":"serif", "color":"darkblue", "size":15}


plt.title("Accuracy Graph", fontdict=font1, loc="left")
plt.xlabel("Epoch", fontdict=font2, loc="center")
plt.ylabel("Loss", fontdict=font2, loc="bottom") 

plt.grid()#ızgara görünümü

yTrainPoints=np.array([0.8,0.75,0.6,0.55,0.45,0.3])
yValPoints=np.array([0.8,0.65,0.6,0.5,0.6,0.7])

plt.plot(yTrainPoints, linestyle=":" , color= "m", lw="2", marker="x")
plt.plot(yValPoints, linestyle="-." , color= "b", lw="4", marker="o")

plt.legend(["Train Loss", "Val Loss"], loc=0)

plt.show()
```

```python
yTrainPoints=np.array([0.8,0.75,0.6,0.55,0.45,0.3])
plt.subplot(1,3,1)
plt.plot(yTrainPoints)
plt.title("Train Graph")
plt.xlabel("Epoch")
plt.ylabel("Loss")

yValPoints=np.array([0.8,0.65,0.6,0.5,0.6,0.7])
plt.subplot(1,3,2)
plt.plot(yValPoints)
plt.title("Val Graph")
plt.xlabel("Epoch")
plt.ylabel("Loss")

yTestPoints=np.array([0.65,0.15,0.35,0.65,0.45,0.35])
plt.subplot(1,3,3)
plt.plot(yTestPoints)
plt.title("Test Graph")
plt.xlabel("Epoch")
plt.ylabel("Loss")
```

```python
x=np.array([rnd(1,100) for i in range(100)])
y=np.array([rnd(1,100) for i in range(100)])

colors=np.array([rnd(1,100) for i in range(100)])
sizes=np.array("[rnd(1,100) for i in range(100)]")
plt.scatter(x,y, c=colors, s=sizes, alpha=0.5,   #alpha: şefaflık saglar noktalarada 
            cmap='GnBu')
plt.colorbar() #bir grafik üzerinde renk sakalasını gözükmesini saglar

plt.show()
```
